# Workday

**`https://{tenant}.{instance}.myworkdayjobs.com/wday/cxs/{tenant}/{site}/jobs`**
* CXS endpoint. Anything served is active posting.

````python
!curl -s 'https://nvidia.wd5.myworkdayjobs.com/wday/cxs/nvidia/NVIDIAExternalCareerSite/jobs' \
  -H 'Content-Type: application/json' \
  -H 'Accept: application/json' \
  --data '{"appliedFacets":{},"limit":20,"offset":0,"searchText":"software engineer intern"}' \
  | jq '.jobPostings[] | {title, locationsText, postedOn, externalPath}'
````

API Endpoint: ```{base}/{locale}/{site} + externalPath```

# Oracle/Taleo

````python
!curl -sS -G 'https://ebxr.fa.us2.oraclecloud.com/hcmRestApi/resources/latest/recruitingCEJobRequisitions' \
  --data-urlencode 'onlyData=true' \
  --data-urlencode 'expand=requisitionList.workLocation,requisitionList.secondaryLocations' \
  --data-urlencode 'finder=findReqs;siteNumber=CX_1,limit=25,offset=0,sortBy=POSTING_DATES_DESC' \
  -H 'ora-irc-language: en' \
  -H "ora-irc-cx-userid: $(uuidgen)" \
  -H 'Content-Type: application/vnd.oracle.adf.resourceitem+json;charset=utf-8' \
  | jq '.items[0].requisitionList[0]'
````
